```python
#First NUmpy code

import numpy as np

# Create a simple 1D array
temperatures = np.array([32.5, 28.9, 30.1, 35.6, 27.3])

print('Array:', temperatures)
print('Data type:', temperatures.dtype)
print('Shape:', temperatures.shape)
print('Average temperature:', temperatures.mean())
print('Maximum temperature:', temperatures.max())
print('Minimum temperature:', temperatures.min())
```
